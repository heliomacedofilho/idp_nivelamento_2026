### Enunciado da Questão 5

Um analista de sistemas foi encarregado de integrar duas plataformas de uma grande empresa. 

A primeira plataforma gera dados macroeconômicos em formato JSON, enquanto a segunda é um sistema legado que só aceita dados em formato XML. 

Ambos os sistemas, porém, usam exatamente os mesmos dados. 

Para garantir que o sistema legado consiga processar corretamente as informações, o analista precisa realizar a conversão dos dados entre os formatos.

Considere o seguinte JSON, que contém informações macroeconômicas do Brasil para o ano de 2023:

```json
{ 
    "pais": "Brasil", 
    "ano": 2023, 
    "indicadores": { 
        "PIB": { 
            "valor": 3.1, 
            "unidade": "trilhões USD" 
        }, 
        "inflacao": { 
            "valor": 4.5, 
            "unidade": "percentual" 
        }, 
        "desemprego": { 
            "valor": 8.2, 
            "unidade": "percentual" 
        } 
    } 
}
```

Construa uma estrutura XML correta e completa, exatamente equivalente, que represente essas mesmas informações
macroeconômicas, de forma que o sistema legado possa processá-las. 

Além de manter a hierarquia exata, todos os pares campos-valor devem ser mapeados em elementos e conteúdos. 

Todas essas informações, no XML, devem estar dentro do elemento raiz `<macroeconomia>`.

👉 dica: use a biblioteca `dict2xml`, realizando a instalação dela na célula abaixo.

```python
from dict2xml import dict2xml

conteudo_json = { 
    "pais": "Brasil", 
    "ano": 2023, 
    "indicadores": { 
        "PIB": { 
            "valor": 3.1, 
            "unidade": "trilhões USD" 
        }, 
        "inflacao": { 
            "valor": 4.5, 
            "unidade": "percentual" 
        }, 
        "desemprego": { 
            "valor": 8.2, 
            "unidade": "percentual" 
        } 
    } 
}

conteudo_xml = dict2xml(conteudo_json, wrap ='macroeconomia')

print(conteudo_xml)
```
