#### CSV

Valores separados por vírgula (CSV, do inglês Comma-separated values) é um formato de arquivo de texto que usa
- vírgulas para separar valores; e 
- quebras de linha para separar registros. 

Um arquivo CSV armazena dados tabulares (números e texto) em texto simples, onde cada linha do arquivo normalmente representa um registro de dados.

Cada registro consiste no mesmo número de campos, e estes são separados por vírgulas no arquivo CSV.

🚨 Se o próprio delimitador de campo puder aparecer dentro de um campo, os campos podem ser cercados por aspas. 🚨

👉 Saiba mais sobre arquivos <a href='https://en.wikipedia.org/wiki/Comma-separated_values'>CSV</a>

<pre>Vamos ler as linhas de um arquivo CSV.</pre>

````python
with open('data/estados.csv') as estados_csv:
    for linha in estados_csv.readlines()[:5]:
        print(linha, end='')
````

<pre>🚨 pandas é simples e amigável para ler csv</pre>

````python
from pandas import read_csv

df = read_csv('data/estados.csv')
df.head()
````

<pre>🚨 igualmente simples e amigável para escrever csv</pre>

````python
df.to_csv('data/output/estados_aux.csv', index=False)
````

#### JSON

Um formato que contém dois tipos de dados:
- Tipos de Dados Primitidos: Números, Cadeias de Caracteres, Booleanos, Nulo.
- Estruturas de Dados: Vetor, Objeto.

👉 Saiba mais sobre arquivos <a href='https://en.wikipedia.org/wiki/JSON'>JSON</a>

Os tipos de dados primitivos são representados como:
- Número: um número decimal (com sinal), que pode conter uma parte fracionária e pode usar a notação exponencial $e$.
    - <i>e.g.</i>, +10.59
- Cadeia de Caracteres: uma sequência de zero ou mais caracteres <a href="https://en.wikipedia.org/wiki/Unicode">unicode</a>, entre aspas duplas.
    - <i>e.g.</i>, "cadeia de caracteres"
- Booleano: ou verdadeiro (<b>true</b>) ou falso (<b>false</b>).
- Nulo: um valor vazio, codificado como <b>null</b>.

As estruturas de dados são representadas como:
- Vetor: uma lista ordenada de zero ou mais elementos, cada um sendo de qualquer tipo de dados.
    - <i>e.g.</i>, [elem1, elem2, elem3]
- Objeto: uma coleção de pares nome-valor, onde os nomes são cadeias de caracteres.
    - <i>e.g.</i>, {"chave1": elem1, "chave2": elem2, "chave3": elem3}

<pre>Vamos ler as linhas de um arquivo JSON.</pre>

````python
with open('data/estados.json') as estados_json:
    estados_json = json.load(estados_json)
    estados_dump = json.dumps(estados_json, indent=2)
    print(estados_dump[:285], end='')
````

<pre>🚨 pandas é simples e amigável para ler json</pre>

````python
from pandas import read_json

df = read_json('data/estados.json')
df.head()
````

<pre>🚨 igualmente é simples e amigável para escrever json</pre>

````python
df.to_json('data/output/estados_aux.json', index=False)
````

#### XML

Formato vem caindo em desuso. Vamos comentar o formato com nosso exemplo concreto.

👉 Saiba mais sobre arquivos <a href='https://en.wikipedia.org/wiki/XML'>XML</a>

<pre>Vamos ler as linhas de um arquivo XML.</pre>

````python
with open('data/estados.xml') as estados_xml:
    estados_xml_str = ''.join(estados_xml.readlines())
    print(estados_xml_str[:530])
````

👉 Caso deseje escrever em abas, precisamos recorrer ao `ExcelWriter`

````python
from pandas import ExcelWriter

with ExcelWriter('data/output/georreferenciamento_aux.xlsx') as writer:
    df_estados.to_excel(writer, sheet_name='Estados', 
                        index=False)
    df_municipios.to_excel(writer, sheet_name='Municípios', 
                           index=False)
````
